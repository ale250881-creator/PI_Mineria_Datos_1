# 05 — Conclusiones

Este notebook responde a las preguntas/objetivos definidos en `03_eda.ipynb`, distinguiendo
**evidencia** (lo que mostraron los datos), **interpretación** (qué significa en relación a la
pregunta) y **conclusión** (qué se puede afirmar, y con qué alcance). Se mantiene coherente con
`reports/informe_final.pdf` y con la app de Streamlit.


## 1. ¿Cómo se distribuye la edad de los usuarios?

- **Evidencia:** media 33.5 años, mediana 33, desvío 11.7, asimetría 0.154 (Q1≈25, Q3≈41).
  Pico aislado en `age = 13` (385 casos, 5 veces más que cualquier edad vecina).
- **Interpretación:** la distribución central es simétrica y describe una audiencia
  mayoritariamente adulta joven/adulta. El pico en 13 años no encaja en ese patrón general:
  es un valor demasiado repetido para ser una coincidencia, pero está dentro de un rango válido,
  por lo que no se puede determinar si es un error de carga o una segmentación real (cuentas
  familiares a nombre de un menor, por ejemplo).
- **Conclusión:** la audiencia se concentra en adultos jóvenes/adultos (~25 a 41 años), con una
  reserva: el pico en `age = 13` queda como hallazgo sin explicación confirmada, no como parte
  de la conclusión sobre el perfil general de edad.


## 2. ¿Qué plan de suscripción predomina?

- **Evidencia:** Básico 45.00%, Estándar 35.21%, Premium 19.79%.
- **Interpretación:** la base de usuarios está concentrada en el plan de entrada, no en el de
  mayor valor por suscripción.
- **Conclusión:** la mayoría de los usuarios paga el plan más económico. Cualquier decisión
  comercial que asuma una base mayoritariamente premium estaría partiendo de una lectura
  incorrecta de esta plataforma.


## 3. ¿El plan de suscripción está asociado a mayor consumo?

- **Evidencia:** mediana de consumo mensual — Básico 553 min, Estándar 840 min, Premium 1.127
  min. Relación confirmada en PCA: PC1 (35.6% de la varianza) está dominado por
  `monthly_watch_time_mins` y `plan_ordinal` con el mismo signo (cargas 0.71 y 0.71).
- **Interpretación:** la relación es ordenada y se sostiene con dos técnicas distintas (EDA y
  PCA): a mayor plan, mayor consumo, de forma consistente.
- **Conclusión:** `subscription_plan` es la variable que mejor explica el nivel de consumo en
  este dataset — el hallazgo más sólido de todo el análisis.


## 4. ¿Más tickets de soporte se asocian a menor consumo?

- **Evidencia:** correlación entre `customer_support_tickets` y `monthly_watch_time_mins` ≈
  -0.0016. El consumo promedio se mantiene estable (~800 min) en todos los niveles de tickets.
- **Interpretación:** no hay evidencia de relación entre ambas variables.
- **Conclusión:** la hipótesis de que más problemas de soporte implican menor consumo **no se
  sostiene** con estos datos. Se documenta como resultado válido, no como un hallazgo fallido.


## 5. ¿El género favorito varía según el país?

- **Evidencia:** las proporciones de cada género se mantienen parecidas en los 7 países
  (diferencias menores a 3 puntos porcentuales en casi todos los casos); Comedia es el género
  más elegido en todos ellos (16% a 19%).
- **Interpretación:** el gusto por género parece transversal a la región, no específico de
  ningún país.
- **Conclusión:** con la evidencia disponible, no se justifica una curaduría de contenido
  diferenciada por país en base al género favorito declarado.


## 6. ¿La relación edad-consumo cambia según el plan?

- **Evidencia:** en el scatter age vs. consumo, los tres planes forman bandas horizontales
  separadas en todas las edades; la correlación edad-consumo dentro de cada plan es ≈0. En PCA,
  PC2 (25.2%) combina `age` y `customer_support_tickets`, pero esa combinación no refleja una
  correlación real entre ambas (su correlación original también es ≈0).
- **Interpretación:** la edad no aporta estructura al comportamiento de consumo, ni sola ni
  combinada con los tickets de soporte.
- **Conclusión:** el plan de suscripción explica el consumo; la edad, no. Esto se confirmó con
  dos técnicas independientes (EDA y PCA), lo que refuerza la solidez de la conclusión.


## Limitaciones

- El pico anómalo en `age = 13` (385 casos) no pudo explicarse con la información disponible:
  permanece como una limitación de calidad de datos, no como un hallazgo del análisis.
- `last_login_date` retiene ~4% de valores faltantes sin imputar (decisión tomada en
  `02_calidad_y_limpieza.ipynb`): no fue posible incorporar análisis de recencia de actividad
  para esos usuarios.
- El diagnóstico MCAR/MAR/MNAR solo pudo contrastarse contra dos variables observadas
  (`subscription_plan` y `country`); no puede descartarse por completo que alguna de las
  variables tratadas como MCAR dependa de otro factor no incluido en el dataset.
- El PCA se realizó únicamente sobre variables numéricas u ordinales; `country` y
  `favorite_genre` (nominales, sin orden natural) quedaron fuera de esa técnica.
- El alcance de las conclusiones se encuentra condicionado por la información disponible y por
  las decisiones documentadas durante el proceso.


## Mejoras futuras

- Indagar con el origen del dataset el motivo del pico en `age = 13`, para decidir si
  corresponde corregirlo, segmentarlo o conservarlo como está.
- Incorporar variables temporales (fecha de alta, historial de actividad) que permitan analizar
  evolución y recencia, no solo una foto fija del estado actual de cada usuario.
- Una mejora futura podría consistir en incorporar información adicional que permita ampliar el
  alcance del análisis — por ejemplo, motivo de los tickets de soporte o dispositivo de
  reproducción.
- Explorar técnicas de reducción de dimensionalidad para variables categóricas (como MCA) si se
  quisiera incorporar `country` y `favorite_genre` a un análisis conjunto con las numéricas.


## Síntesis final

El hallazgo central y mejor respaldado de todo el proyecto es que **`subscription_plan`
organiza el comportamiento de consumo** de los usuarios — confirmado de forma independiente
en el EDA (Pregunta 3) y en PCA (PC1, 35.6% de la varianza). Dos hipótesis intuitivas se
pusieron a prueba y **no se sostuvieron**: ni los tickets de soporte ni la edad explican el
consumo. Mostrar estos resultados negativos con el mismo nivel de evidencia que los positivos
es, en sí mismo, parte de la conclusión del proyecto.
